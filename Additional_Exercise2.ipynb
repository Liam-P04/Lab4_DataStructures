```python
from collections import Counter, defaultdict
from statistics import mean

LAST_NAME = "Pabiona"
STUDENT_ID = "TUPM-26-4160"
FAVORITE_ARTIST = input("Enter your favorite artist: ").strip()
while not FAVORITE_ARTIST:
    FAVORITE_ARTIST = input("Please enter a non-empty favorite artist: ").strip()

SEED_NUM = int(STUDENT_ID[-1])
STUDENT_NAMES = [f"{LAST_NAME}_Student_{index}" for index in range(1, 6)]
SUBJECTS = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]


def performance_category(score):
    if score is None:
        return "INCOMPLETE"
    if score >= 90:
        return "EXCELLENT"
    if score >= 80:
        return "GOOD"
    if score >= 70:
        return "SATISFACTORY"
    if score >= 60:
        return "NEEDS IMPROVEMENT"
    return "FAILING"


records = []
for student_index, student in enumerate(STUDENT_NAMES, start=1):
    for subject_index, subject in enumerate(SUBJECTS, start=1):
        artist_position = (SEED_NUM + student_index * 2 + subject_index - 1) % len(FAVORITE_ARTIST)
        artist_character = FAVORITE_ARTIST[artist_position]
        score = 35 + (
            SEED_NUM + (artist_position + 1) * 11 + student_index * 7 + subject_index * 13
        ) % 66
        records.append({
            "student": student,
            "subject": subject,
            "score": score,
            "category": performance_category(score),
            "artist_position": artist_position + 1,
            "artist_character": artist_character,
        })

records[-1]["score"] = None
records[-1]["category"] = performance_category(None)
records.append(records[0].copy())

record_counts = Counter(
    (record["student"], record["subject"], record["score"])
    for record in records
)
repeated_records = [
    {"student": student, "subject": subject, "score": score, "occurrences": count}
    for (student, subject, score), count in record_counts.items()
    if count > 1
]
incomplete_records = [record for record in records if record["score"] is None]

unique_records = []
seen_record_keys = set()
for record in records:
    record_key = tuple(sorted(record.items()))
    if record_key not in seen_record_keys:
        unique_records.append(record)
        seen_record_keys.add(record_key)

scores_by_student = defaultdict(list)
scores_by_subject = defaultdict(list)
categories_by_score = Counter()
for record in unique_records:
    if record["score"] is not None:
        scores_by_student[record["student"]].append(record["score"])
        scores_by_subject[record["subject"]].append(record["score"])
        categories_by_score[record["category"]] += 1

ordered_records = sorted(
    records,
    key=lambda record: (
        record["student"],
        record["subject"],
        record["score"] is None,
        record["score"] if record["score"] is not None else 0,
    ),
)

print("=== GENERATED PERFORMANCE RECORDS ===")
print(f"Student: {LAST_NAME} | Seed: {SEED_NUM} | Favorite artist: {FAVORITE_ARTIST}")
print(f"Score bands: Excellent 90-100, Good 80-89, Satisfactory 70-79, "
      "Needs Improvement 60-69, Failing below 60")
for record in records:
    print(record)

print("\n=== REPEATED RECORDS ===")
if repeated_records:
    for record in repeated_records:
        print(record)
else:
    print("None")

print("\n=== INCOMPLETE RECORDS ===")
if incomplete_records:
    for record in incomplete_records:
        print(record)
else:
    print("None")

print("\n=== PERFORMANCE CATEGORIES ===")
for category, count in sorted(categories_by_score.items()):
    print(f"{category}: {count}")
print(f"INCOMPLETE: {len(incomplete_records)}")

print("\n=== STUDENT SUMMARIES ===")
for student in STUDENT_NAMES:
    scores = scores_by_student[student]
    print(
        f"{student}: completed={len(scores)}, "
        f"average={mean(scores):.2f}, highest={max(scores)}, lowest={min(scores)}"
    )

print("\n=== SUBJECT SUMMARIES ===")
for subject in SUBJECTS:
    scores = scores_by_subject[subject]
    print(
        f"{subject}: completed={len(scores)}, "
        f"average={mean(scores):.2f}, highest={max(scores)}, lowest={min(scores)}"
    )

print("\n=== ORDERED RESULTS ===")
for record in ordered_records:
    print(record)

print("\n=== CONSOLIDATED REPORT ===")
print(
    f"{len(records)} total rows, {len(unique_records)} distinct records, "
    f"{sum(record['score'] is not None for record in unique_records)} distinct scored records, "
    f"{len(repeated_records)} repeated record group(s), and "
    f"{len(incomplete_records)} incomplete record(s)."
)
```
