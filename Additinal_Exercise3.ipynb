```python
from collections import Counter, defaultdict

LAST_NAME = "Pabiona"
SEED_NUM = 7
FAVORITE_ARTIST = input("Enter your favorite artist: ").strip()
while not FAVORITE_ARTIST:
    FAVORITE_ARTIST = input("Please enter a non-empty favorite artist: ").strip()

CUSTOMERS = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
PRODUCTS = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
PRICES = {
    "NOTEBOOK": 150.00,
    "PEN": 20.00,
    "USB": 350.00,
    "HEADPHONES": 800.00,
    "CALCULATOR": 600.00,
}

records = []
for customer_index, customer in enumerate(CUSTOMERS, start=1):
    for product_index, product in enumerate(PRODUCTS):
        artist_position = (SEED_NUM + customer_index + product_index) % len(FAVORITE_ARTIST)
        artist_char = FAVORITE_ARTIST[artist_position]
        quantity = (SEED_NUM + customer_index + product_index + ord(artist_char)) % 5 + 1

        transaction = {
            "customer": customer,
            "product": product,
            "quantity": quantity,
            "unit_price": PRICES[product],
            "transaction_value": round(quantity * PRICES[product], 2),
            "artist_position": artist_position + 1,
            "artist_char": artist_char,
        }
        records.append(transaction)

# Repeated purchase: ensure there is at least one repeated customer-product transaction.
records.append(records[0].copy())

# Organize records by customer, product, and total value.
by_customer = defaultdict(list)
by_product = defaultdict(list)
by_customer_product = defaultdict(float)

for record in records:
    by_customer[record["customer"]].append(record)
    by_product[record["product"]].append(record)
    by_customer_product[(record["customer"], record["product"])] += record["transaction_value"]

record_keys = [(record["customer"], record["product"], record["quantity"], record["unit_price"]) for record in records]
record_counts = Counter(record_keys)
repeated_records = [
    {"customer": customer, "product": product, "quantity": quantity, "unit_price": price, "occurrences": count}
    for (customer, product, quantity, price), count in record_counts.items()
    if count > 1
]

distinct_customers = sorted({record["customer"] for record in records})
distinct_products = sorted({record["product"] for record in records})
total_transaction_value = sum(record["transaction_value"] for record in records)

print("=== GENERATED TRANSACTION DATA ===")
print(f"Student: {LAST_NAME} | Seed: {SEED_NUM} | Favorite artist: {FAVORITE_ARTIST}")
for record in records:
    print(record)

print("\n=== REPEATED & DISTINCT DATA ===")
print(f"Total records: {len(records)}")
print(f"Distinct customers: {distinct_customers}")
print(f"Distinct products: {distinct_products}")
print(f"Repeated purchase groups: {repeated_records}")

print("\n=== CUSTOMER SUMMARY ===")
for customer in sorted(by_customer):
    customer_total = sum(record["transaction_value"] for record in by_customer[customer])
    print(f"{customer}: total value = {customer_total:.2f}")

print("\n=== PRODUCT SUMMARY ===")
for product in sorted(by_product):
    product_total = sum(record["transaction_value"] for record in by_product[product])
    print(f"{product}: total value = {product_total:.2f}")

print("\n=== CUSTOMER-PRODUCT SUMMARY ===")
for (customer, product), value in sorted(by_customer_product.items()):
    print(f"{customer} - {product}: {value:.2f}")

ordered_records = sorted(records, key=lambda record: (record["customer"], record["product"], record["transaction_value"]))
print("\n=== ORDERED TRANSACTION REPORT ===")
for record in ordered_records:
    print(
        f"{record['customer']} | {record['product']} | qty={record['quantity']} | "
        f"unit_price={record['unit_price']:.2f} | total={record['transaction_value']:.2f}"
    )

print("\n=== FINAL CONSOLIDATED TRANSACTION REPORT ===")
print(f"Total transaction value: {total_transaction_value:.2f}")
print(f"Average transaction value: {total_transaction_value / len(records):.2f}")
print(f"Repeated purchase entries: {len(repeated_records)}")
print(f"Customer count: {len(distinct_customers)}")
print(f"Product count: {len(distinct_products)}")
```
