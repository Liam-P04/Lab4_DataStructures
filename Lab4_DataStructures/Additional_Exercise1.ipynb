```python
from collections import Counter, defaultdict
from statistics import mean, median

LAST_NAME = "Pabiona"
STUDENT_ID = "TUPM-26-4160"
FAVORITE_ARTIST = input("Enter your favorite artist: ").strip()
while not FAVORITE_ARTIST:
    FAVORITE_ARTIST = input("Please enter a non-empty favorite artist: ").strip()

SEED_NUM = int(STUDENT_ID[-1])
TEST_LABELS = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
CATEGORIES = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

records = []
for test_index, test_source in enumerate(TEST_LABELS, start=1):
    for category_index, category in enumerate(CATEGORIES):
        artist_position = (SEED_NUM + test_index + category_index - 2) % len(FAVORITE_ARTIST)
        artist_character = FAVORITE_ARTIST[artist_position]
        value = SEED_NUM * 10 + ord(artist_character) + test_index * 3 + category_index
        records.append({
            "test_source": test_source,
            "category": category,
            "value": value,
            "artist_position": artist_position + 1,
            "artist_character": artist_character,
        })

records.append(records[0].copy())

record_keys = [tuple(sorted(record.items())) for record in records]
record_counts = Counter(record_keys)
repeated_records = [
    {"record": dict(record_key), "occurrences": count}
    for record_key, count in record_counts.items()
    if count > 1
]
distinct_records = [dict(record_key) for record_key in record_counts]

by_source = defaultdict(list)
by_category = defaultdict(list)
for record in records:
    by_source[record["test_source"]].append(record)
    by_category[record["category"]].append(record["value"])

print("=== GENERATED TEST DATA ===")
print(f"Student: {LAST_NAME} | Seed: {SEED_NUM} | Favorite artist: {FAVORITE_ARTIST}")
for record in records:
    print(record)

print("\n=== REPEATED AND DISTINCT DATA ===")
print(f"Total records: {len(records)}")
print(f"Distinct records: {len(distinct_records)}")
print(f"Repeated record groups: {len(repeated_records)}")
for item in repeated_records:
    print(f"Repeated {item['occurrences']} times: {item['record']}")

print("\n=== NUMERICAL SUMMARIES BY CATEGORY ===")
for category in CATEGORIES:
    values = by_category[category]
    print(
        f"{category}: count={len(values)}, min={min(values)}, max={max(values)}, "
        f"mean={mean(values):.2f}, median={median(values):.2f}"
    )

print("\n=== CATEGORICAL SUMMARIES ===")
print("Measurements per category:", {category: len(by_category[category]) for category in CATEGORIES})
print("Measurements per test source:", {source: len(by_source[source]) for source in TEST_LABELS})

ordered_records = sorted(records, key=lambda record: (record["test_source"], record["category"], record["value"]))
print("\n=== ORDERED PROCESSED RESULTS ===")
for record in ordered_records:
    print(record)

print("\n=== FINAL SUMMARY ===")
print(
    f"Processed {len(records)} records from {len(TEST_LABELS)} test sources "
    f"across {len(CATEGORIES)} categories; {len(distinct_records)} distinct records "
    f"and {sum(count - 1 for count in record_counts.values())} extra repeated record(s)."
)
```
